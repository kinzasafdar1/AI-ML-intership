```python
df = df.drop_duplicates()

print("Dataset shape:", df.shape)
```

```python
df = df.drop(columns=["customer_id"])
```

```python
df["SingleProfile"] = (df["number_of_profiles"] == 1).astype(int)

print(
    df[["number_of_profiles", "SingleProfile"]].head()
)
```

```python
# Engineered features (skip any you didn't create earlier)
df["ManyProfiles"] = (df["number_of_profiles"] >= 4).astype(int)
df["WatchHoursPerProfile"] = df["watch_hours"] / df["number_of_profiles"]
df["Inactive30"] = (df["last_login_days"] > 30).astype(int)

X = df[
    ["age", "gender", "subscription_type", "watch_hours",
     "last_login_days", "region", "device", "payment_method",
     "number_of_profiles", "avg_watch_time_per_day", "favorite_genre",
     "ManyProfiles", "WatchHoursPerProfile", "Inactive30"]
]

y = df["churned"]

print("Features:")
print(X.head())

print("\nTarget:")
print(y.head())

print("\nShapes:", X.shape, y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training rows:", X_train.shape[0])
print("Testing rows:", X_test.shape[0])

print("Training class proportions:")
print(y_train.value_counts(normalize=True))

print("Testing class proportions:")
print(y_test.value_counts(normalize=True))
```

```python
numeric_features = [
    "age", "watch_hours", "last_login_days", "number_of_profiles",
    "avg_watch_time_per_day", "WatchHoursPerProfile",
    "ManyProfiles", "Inactive30"
]

categorical_features = [
    "gender", "subscription_type", "region",
    "device", "payment_method", "favorite_genre"
]
```

```python
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
```

```python
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore",
                              sparse_output=False))
])
```

```python
preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])
```

```python
X_train_prepared = preprocessor.fit_transform(X_train)

X_test_prepared = preprocessor.transform(X_test)

print("Training shape:", X_train_prepared.shape)
print("Testing shape:", X_test_prepared.shape)
```

```python
print("Missing values in training:",
      np.isnan(X_train_prepared).sum())

print("Missing values in testing:",
      np.isnan(X_test_prepared).sum())

print("Training target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
    classification_report
)

import matplotlib.pyplot as plt
import pandas as pd
```

```python
logistic_model = LogisticRegression(max_iter=1000)

logistic_model.fit(X_train_prepared, y_train)

logistic_predictions = logistic_model.predict(X_test_prepared)

print("First 10 predictions:", logistic_predictions[:10])
print("First 10 actual values:", y_test.iloc[:10].tolist())
```

```python

knn_model = KNeighborsClassifier(n_neighbors=5)

knn_model.fit(X_train_prepared, y_train)

knn_predictions = knn_model.predict(X_test_prepared)

print("First 10 KNN predictions:", knn_predictions[:10])
```

```python
tree_model = DecisionTreeClassifier(
    max_depth=5,
    random_state=42
)

tree_model.fit(X_train_prepared, y_train)

tree_predictions = tree_model.predict(X_test_prepared)

print("First 10 tree predictions:", tree_predictions[:10])
```

```python
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)
import matplotlib.pyplot as plt


def evaluate_model(name, y_true, y_pred):

    print("\nModel:", name)

    print("Accuracy:",
          accuracy_score(y_true, y_pred))

    print("F1-score:",
          f1_score(y_true, y_pred, zero_division=0))

    print("\nClassification report:")
    print(classification_report(
        y_true,
        y_pred,
        target_names=["Stayed", "Churned"],
        zero_division=0
    ))

    cm = confusion_matrix(y_true, y_pred)

    display = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=["Stayed", "Churned"]
    )

    display.plot()
    plt.title(name + " Confusion Matrix")
    plt.show()


evaluate_model("Logistic Regression", y_test, logistic_predictions)
```

```python
evaluate_model("Logistic Regression",
               y_test, logistic_predictions)

evaluate_model("KNN",
               y_test, knn_predictions)

evaluate_model("Decision Tree",
               y_test, tree_predictions)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.metrics import mean_absolute_error, r2_score

watch_features = [
    "age", "number_of_profiles", "last_login_days",
    "Inactive30", "ManyProfiles"
]

X_watch_train = X_train[watch_features]
X_watch_test = X_test[watch_features]

y_watch_train = X_train["watch_hours"]
y_watch_test = X_test["watch_hours"]

watch_model = make_pipeline(
    SimpleImputer(strategy="median"),
    LinearRegression()
)

watch_model.fit(X_watch_train, y_watch_train)

watch_predictions = watch_model.predict(X_watch_test)

print("Actual watch hours:", y_watch_test.head().tolist())
print("Predicted watch hours:", watch_predictions[:5])

print("Mean Absolute Error:",
      mean_absolute_error(y_watch_test, watch_predictions))
print("R2:", r2_score(y_watch_test, watch_predictions))
```

```python
results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "KNN",
        "Decision Tree"
    ],
    "Accuracy": [
        accuracy_score(y_test, logistic_predictions),
        accuracy_score(y_test, knn_predictions),
        accuracy_score(y_test, tree_predictions)
    ],
    "F1-score": [
        f1_score(y_test, logistic_predictions),
        f1_score(y_test, knn_predictions),
        f1_score(y_test, tree_predictions)
    ]
})

print(results.round(3))
```
